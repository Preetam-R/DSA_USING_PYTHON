# **Traversing in linked list**

```python
#first create structure of node using class

class Node:
  def __init__(self,data):
    self.data = data
    self.next = None

#assign the values to the node
node1 = Node(10)
node2 = Node(30)
node3 = Node(50)
node4 = Node(60)

node1.next = node2
node2.next = node3
node3.next = node4

#we now make use of head
head = node1

#generally we dont move the head, instead we use current variable for traversing
current = head

#traversing Linked List using While Loop
while current.next is not None:
  print(current.data,end = "->")
  current = current.next
print("NUll")

#this is diplay function, in which current is the variable and traverse throught the linked list
```
