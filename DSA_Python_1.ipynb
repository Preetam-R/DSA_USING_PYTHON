## **Stacks**

```python
import sys

#stacks using oops

class Stack:
  def __init__(self):
    self.stack = []

  def push(self,element):
    self.stack.append(element)

  def isempty(self):
    return len(self.stack) == 0


  def pop(self):
    if not self.isempty(): # Pop only if the stack is NOT empty
      return self.stack.pop()
    else:
      return "Stack is empty, cannot pop."

  def display(self):
    if not self.isempty(): # Display elements only if the stack is NOT empty
      for i in self.stack:
        print(i, end = " ")
      print() # Add a newline for better formatting
    else:
      print("Stack is empty.")


st = Stack()

while True:
  print("\n1.Push\n2.Pop\n3.Display\n4.Exit")
  try:
    ch = int(input("Enter your choice: "))

    if ch == 1:
      r = int(input("Enter the number of elements to push: "))
      for i in range(r):
        try:
          ele = int(input(f"Enter element {i+1}: "))
          st.push(ele)
        except ValueError:
          print("Invalid input. Please enter an integer.")
    elif ch == 2:
      popped_element = st.pop()
      if isinstance(popped_element, str):
        print(popped_element)
      else:
        print(f"Popped element: {popped_element}")
    elif ch == 3:
      st.display()
    elif ch == 4:
      print("Exiting program.")
      break
    else:
      print("Enter a choice between 1 and 4")
  except ValueError:
    print("Invalid input. Please enter a number for your choice.")
```

# **QUEUE**

```python
class Queue:
  def __init__(self):
    self.queue = []

  def enqueue(self):
    rge = int(input("enter the no. of elements."))
    if rge == 1:
      element = int(input("enter the element"))
      self.queue.append(element)
    else:
      for i in range(1,rge+1):
        element = int(input(f"enter the {i} element."))
        self.queue.append(element)

  def isempty(self):
    return len(self.queue) == 0

  def dequeue(self):
    if self.isempty():
      print("Queue is emptY!!!")
    else:
      self.queue.pop(0)

  def peek(self):
    return self.queue[0]

  def display(self):
    if not self.isempty():
      print("the elements of queue are:")
      for i in self.queue:
        print(i,end = " ")
    else:
      print("queue is empty!!!! push some elements in the queue")


q = Queue()

while True:
  print()
  print("\n1.ENQUEUE\n 2.DEQUEUE\n 3.DISPLAY\n 4.EXIT")

  ch = int(input("enter you choice:"))

  if ch == 1:
    q.enqueue()
  elif ch == 2:
    q.dequeue()
  elif ch == 3:
    q.display()
  elif ch == 4:
    print("exiting program...")
    break
  else:
    print("enter the choice between 1 and 4")








```
