# **Inserting Node at Beginning**

```python

class Node:
  def __init__(self,data):
    self.data = data
    self.next = None

#creating the Nodes

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)

#connecting the nodes

node1.next = node2
node2.next = node3
node3.next = node4

head = node1

current = head

#creating the new node
new_node = Node(69)
new_node.next = head  #linking new node to head
head = new_node  #assigning new node as head


while current is not None:
  print(current.data,end = "->")
  current = current.next
print("NULL")


```
