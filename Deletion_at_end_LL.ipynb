```python
class Node:
  def __init__(self,data):
    self.data = data
    self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1
current = head
print("\nbefore deleting node at end")
while current is not None:
  print(current.data,end = "->")
  current = current.next
print("NULL")

print("\nafter deleting the node at end")

# Reset current to head before attempting to delete the node
current = head

# Handle the case of an empty list or a single-node list (though the current setup has multiple nodes)
if current is None or current.next is None:
    head = None # List becomes empty or was already empty
else:
    while current.next.next is not None:
        current = current.next
    current.next = None

current = head # Reset current to head for printing the modified list
while current is not None:
  print(current.data,end = "->")
  current = current.next
print("NULL")
```
